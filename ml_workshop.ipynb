```python
for i in ['Customer_Satisfaction_Score','Marketing_Response_Rate','Income']:
  df[i]=df[i].fillna(df[i].median())
```

```python
df['Customer_Satisfaction_Score'].isnull().sum()
```

```python
for j in ['Website_Visits_Monthly','Preferred_Category','App_Sessions_Monthly','Acquisition_Channel']:
  df[j] = df[j].fillna(df[j].mode()[0])
```

```python
df['Income'].isnull().sum()
```

```python
X = df[['Income']]
y= df['Annual_Revenue']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.25,random_state=11)
```

```python
print(X_train.shape,X_test.shape,y_train.shape,y_test.shape)
```

```python
X_test.head()
```

```python
from sklearn.linear_model import LinearRegression
model_1 = LinearRegression()
```

```python
model_1.fit(X_train,y_train)
```

```python
y_pred = model_1.predict(X_test)
```

```python
print(y_test,y_pred)
```

```python
from sklearn.metrics import r2_score
r2_score(y_test,y_pred)
```

```python
X = df[["Income",'Website_Visits_Monthly',"Customer_Tenure_Months","Purchase_Frequency_Monthly","Average_Order_Value"]]
y = df['Annual_Revenue']

from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.25,random_state=11)
model_1 = LinearRegression()
model_1.fit(X_train,y_train)
y_pred = model_1.predict(X_test)
print(r2_score(y_test,y_pred))
```
